### Stats libary implementation of Correlation Calculation:

```python
# H0: No relation among two variable
# HA: relationship among two variable

from scipy.stats import pearsonr
from scipy.stats import spearmanr


x = [1, 3, 4, 5, 2]
y = [1, 2, 5, 4, 3]

p_corr, p_value = pearsonr(x, y)
print("pearson Corr", p_corr)
print("p_value", p_value)
```

```python

s_corr, s_value = spearmanr(x, y)
print("Spearmann Corr", s_corr)
print("p_value", p_value)
```

```python
p_corr, p_value = pearsonr(df_hw['Height'], df_hw['Weight'])
print("pearson Corr", p_corr)
print("p_value", p_value)
```

```python
s_corr, p_value = spearmanr(df_hw['Height'], df_hw['Weight'])
print("spearmanr Corr", s_corr)
print("p_value", p_value)
```

```python
x = [1, 2, 3, 4, 5]
y = [5, 4, 3, 2, 1]
s_corr, s_value = spearmanr(x, y)
print("spearmanr Corr", s_corr)
```

```python
x = [1, 2, 3, 4, 5]
y = [5, 4, 3, 2, 1]
p_corr, p_value = pearsonr(x, y)
print("pearson Corr", p_corr)
```

```python
x = [1, 2, 3, 4, 5, 6, 7]
y = [10, 9, 2.5, 6, 4, 3, 2]

p_corr, p_value = pearsonr(x, y)
print("pearson Corr", p_corr)
print("p_value", p_value)
```
